```python

def find_matching_system(f_center):
    """Finds candidate system from catalog matching frequency center."""
    for sys_name, specs in rfi_specs.items():
        f_start, f_stop = specs["band"]
        if f_start <= f_center <= f_stop:
            return sys_name, specs
    return "Unknown / Unlicensed", None
```

```python

results = []

for idx, row in high_peak_df.iterrows():

    # --------------------------------------------------
    # 1. Measured RFI parameters
    # --------------------------------------------------

    detected_freq = row['detected_frequency_[MHz]']

    measured_obw = row['occupied_bandwidth_[MHz]']
    measured_fwhm = row['fwhm_[MHz]']

    # Measured OBW/FWHM ratio
    measured_ratio = (
        measured_obw / measured_fwhm
        if measured_fwhm > 0
        else np.nan
    )

    # --------------------------------------------------
    # 2. Identify matching system
    # --------------------------------------------------

    system_name, system_specs = find_matching_system(
        detected_freq
    )

    # --------------------------------------------------
    # 3. If no matching system was found
    # --------------------------------------------------

    if system_specs is None:

        results.append({
            'peak_index': idx,
            'detected_frequency_[MHz]': detected_freq,

            'matched_system': 'Unknown',

            'measured_obw_[MHz]': measured_obw,
            'catalog_obw_[MHz]': np.nan,
            'obw_ratio_detected_catalog': np.nan,

            'measured_fwhm_[MHz]': measured_fwhm,
            'catalog_fwhm_[MHz]': np.nan,
            'fwhm_ratio_detected_catalog': np.nan,

            'measured_obw_fwhm_ratio': measured_ratio,
            'catalog_obw_fwhm_ratio': np.nan,

            'obw_ratio_error': np.nan,
            'fwhm_ratio_error': np.nan,
            'shape_ratio_error': np.nan
        })

        continue

    # --------------------------------------------------
    # 4. Catalog parameters
    # --------------------------------------------------

    catalog_obw = system_specs['obw']
    catalog_fwhm = system_specs['fwhm']

    # --------------------------------------------------
    # 5. Compare measured OBW with catalog OBW
    # --------------------------------------------------

    if catalog_obw > 0:

        obw_ratio = (
            measured_obw / catalog_obw
        )

        # Relative error from the ideal ratio of 1
        obw_ratio_error = abs(
            obw_ratio - 1
        )

    else:

        obw_ratio = np.nan
        obw_ratio_error = np.nan

    # --------------------------------------------------
    # 6. Compare measured FWHM with catalog FWHM
    # --------------------------------------------------

    if catalog_fwhm > 0:

        fwhm_ratio = (
            measured_fwhm / catalog_fwhm
        )

        fwhm_ratio_error = abs(
            fwhm_ratio - 1
        )

    else:

        fwhm_ratio = np.nan
        fwhm_ratio_error = np.nan

    # --------------------------------------------------
    # 7. Compare OBW/FWHM ratios
    # --------------------------------------------------

    if catalog_fwhm > 0:

        catalog_ratio = (
            catalog_obw / catalog_fwhm
        )

    else:

        catalog_ratio = np.nan

    if catalog_ratio > 0 and not np.isnan(measured_ratio):

        shape_ratio_error = abs(
            measured_ratio - catalog_ratio
        ) / catalog_ratio

    else:

        shape_ratio_error = np.nan

    # --------------------------------------------------
    # 8. Store result
    # --------------------------------------------------

    results.append({

        'detected_frequency_[MHz]':
            f'{detected_freq:.2f}',

        'matched_system':
            system_name,

        # OBW
        'measured_obw_[MHz]':
            f'{measured_obw:.2f}',

        'catalog_obw_[MHz]':
            f'{catalog_obw:.2f}',

        'obw_ratio_detected_catalog':
            obw_ratio,

        'obw_ratio_error':
            obw_ratio_error,

        # FWHM
        'measured_fwhm_[MHz]':
            f'{measured_fwhm:.2f}',

        'catalog_fwhm_[MHz]':
            f'{catalog_fwhm:.2f}',

        'fwhm_ratio_detected_catalog':
            f'{fwhm_ratio:.2f}',

        'fwhm_ratio_error':
            f'{fwhm_ratio_error:.2f}',

        # Shape
        'measured_obw_fwhm_ratio':
            f'{measured_ratio:.2f}',

        'catalog_obw_fwhm_ratio':
            f'{catalog_ratio:.2f}',

        'shape_ratio_error':
            f'{shape_ratio_error:.2f}'
    })


# ------------------------------------------------------
# 9. Convert results into a dataframe
# ------------------------------------------------------

rfi_evaluation_df = pd.DataFrame(results)

# Optional: round numerical values
rfi_evaluation_df = rfi_evaluation_df.round(4)

rfi_evaluation_df
```
